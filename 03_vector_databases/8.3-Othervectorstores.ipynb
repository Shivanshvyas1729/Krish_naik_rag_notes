### InMemoryVectorStore
In-memory vector store implementation.

Uses a dictionary, and computes cosine similarity for search using numpy.

---

### 💡 Interview & Learning Notes

**Key Interview Questions:**
1. *When should you use `InMemoryVectorStore` vs FAISS vs ChromaDB?* 
   - `InMemoryVectorStore` is the absolute simplest, dependency-free vector store available directly in `langchain-core`. It is perfect for tiny datasets (e.g., embedding 10 tweets or a single small document) or unit testing where you don't want to install heavy C++ libraries like FAISS or databases like Chroma.
2. *Does `InMemoryVectorStore` scale?* 
   - No. It loops through a dictionary and computes NumPy cosine similarities sequentially. It will crash or become unusably slow on datasets larger than a few thousand vectors.

**Learning Takeaways:**
- Use the right tool for the job. Not every RAG pipeline needs a massive scalable vector database. If your context window is small and document count is low, an in-memory dictionary is perfectly fine.

### 🚀 Best Practices for InMemoryVectorStore

1. **Cost Efficiency & Token Optimization**:
   - Extremely cost-efficient for testing LLM Prompts and Chains since it requires zero database infrastructure.

2. **Time Optimization**:
   - Never use this in production for large datasets. `numpy` dot products inside a Python for-loop are drastically slower than compiled indexing algorithms like HNSW (used by FAISS/Chroma).

3. **Data Quality**:
   - Because data is ephemeral, if the Python script crashes, the embeddings are gone. Only use this for static lists or data fetched freshly at runtime.
